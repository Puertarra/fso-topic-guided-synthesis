```python

# Suprimir warnings de deprecación de NumPy para archivos pickle
#warnings.filterwarnings('ignore', category=DeprecationWarning, module='numpy')

def load_dataframes(df_paths: Dict[str, str]) -> Dict[str, pd.DataFrame]:
    """
    Carga todos los dataframes desde los archivos pickle

    Args:
        df_paths: Diccionario con nombres y rutas de los dataframes

    Returns:
        Diccionario con los dataframes cargados
    """
    dataframes = {}
    for name, path in df_paths.items():
        try:
            with open(path, 'rb') as f:
                dataframes[name] = pickle.load(f)
            print(f"✓ Cargado exitosamente: {name}")
        except FileNotFoundError:
            print(f"✗ Archivo no encontrado: {path}")
        except Exception as e:
            print(f"✗ Error cargando {name}: {str(e)}")

    return dataframes

def get_documents_by_topics(dataframes: Dict[str, pd.DataFrame],
                            topic_indices: List[int],
                            supertopic: str = None) -> List[str]:
    """
    Recupera todos los documentos (Concatenated_Info) que pertenecen a los tópicos especificados
    ELIMINA DUPLICADOS DENTRO del mismo supertópico, pero permite duplicación entre supertópicos

    Args:
        dataframes: Diccionario con todos los dataframes cargados
        topic_indices: Lista de índices de tópicos a buscar
        supertopic: Nombre del supertópico (para debugging)

    Returns:
        Lista de strings únicos con la información concatenada de los documentos
    """
    all_documents = []

    for df_name, df in dataframes.items():
        if 'Topic' in df.columns and 'Concatenated_Info' in df.columns:
            filtered_docs = df[df['Topic'].isin(topic_indices)]['Concatenated_Info'].tolist()
            all_documents.extend(filtered_docs)

    # ELIMINAR duplicados SOLO dentro de este supertópico manteniendo el orden
    unique_documents = []
    seen = set()
    for doc in all_documents:
        if doc and doc not in seen:
            unique_documents.append(doc)
            seen.add(doc)

    return unique_documents

def count_tokens(text: str) -> int:
    """
    Cuenta los tokens en un texto (palabras separadas por espacios)

    Args:
        text: Texto a procesar

    Returns:
        Número de tokens
    """
    if pd.isna(text) or text == "":
        return 0
    # Tokenización básica: dividir por espacios y filtrar tokens vacíos
    tokens = re.findall(r'\b\w+\b', str(text).lower())
    return len(tokens)

def create_supertopic_dataframe(questions_dict: Dict[str, List[str]],
                               category_indices: Dict[str, List[int]],
                               df_paths: Dict[str, str]) -> pd.DataFrame:
    """
    Crea un dataframe con supertópicos, preguntas asociadas y documentos correspondientes

    Args:
        questions_dict: Diccionario con preguntas por supertópico
        category_indices: Diccionario con índices de tópicos por supertópico
        df_paths: Rutas de los dataframes

    Returns:
        DataFrame con columnas: supertopic, questions, documents
    """

    print("Cargando dataframes...")
    dataframes = load_dataframes(df_paths)

    if not dataframes:
        print("No se pudieron cargar los dataframes.")
        return pd.DataFrame()

    supertopic_data = []

    print("\nProcesando supertópicos...")
    for supertopic in questions_dict.keys():
        print(f"\nProcesando: {supertopic}")

        # Obtener preguntas del supertópico
        questions = questions_dict[supertopic]

        # Obtener índices de tópicos asociados
        topic_indices = category_indices.get(supertopic, [])
        print(f"  Tópicos asociados: {topic_indices}")

        # Recuperar documentos correspondientes
        documents = get_documents_by_topics(dataframes, topic_indices, supertopic)

        # Agregar información al resultado
        supertopic_data.append({
            'supertopic': supertopic,
            'questions': questions,
            'documents': documents,
            'num_questions': len(questions),
            'num_documents': len(documents)
        })

        print(f"  ✓ {len(questions)} preguntas, {len(documents)} documentos (incluyendo posibles duplicados)")

    # Crear DataFrame
    result_df = pd.DataFrame(supertopic_data)

    print(f"\n🎯 DataFrame creado con {len(result_df)} supertópicos")
    print(f"Total de documentos (con duplicados si aplica): {sum(len(docs) for docs in result_df['documents'])}")

    return result_df

def create_documents_tokens_dataframe(supertopic_df: pd.DataFrame) -> pd.DataFrame:
    """
    Crea un dataframe independiente con cada documento y su conteo de tokens

    Args:
        supertopic_df: DataFrame con supertópicos y documentos

    Returns:
        DataFrame con columnas: document, tokens
    """
    documents_data = []
    document_counter = 1

    print("\nCreando DataFrame de documentos individuales con tokens...")

    for idx, row in supertopic_df.iterrows():
        supertopic = row['supertopic']
        documents = row['documents']

        print(f"Procesando {len(documents)} documentos de: {supertopic}")

        for doc_idx, doc in enumerate(documents, 1):
            if doc and doc.strip():  # Verificar que el documento no esté vacío
                tokens_count = count_tokens(doc)

                documents_data.append({
                    'document_id': f"doc_{document_counter:04d}",
                    'supertopic': supertopic,
                    'document_position_in_supertopic': doc_idx,
                    'document': doc,
                    'tokens': tokens_count,
                    'char_length': len(str(doc))
                })

                document_counter += 1

    documents_df = pd.DataFrame(documents_data)

    if not documents_df.empty:
        print(f"\n📊 DataFrame de documentos creado:")
        print(f"   - Total de documentos: {len(documents_df)}")
        print(f"   - Total de tokens: {documents_df['tokens'].sum():,}")
        print(f"   - Promedio de tokens por documento: {documents_df['tokens'].mean():.1f}")
        print(f"   - Documento más largo: {documents_df['tokens'].max():,} tokens")
        print(f"   - Documento más corto: {documents_df['tokens'].min():,} tokens")

        # Mostrar distribución por supertópico
        print(f"\n📈 Distribución por supertópico:")
        for supertopic in documents_df['supertopic'].unique():
            subset = documents_df[documents_df['supertopic'] == supertopic]
            print(f"   - {supertopic}: {len(subset)} docs, {subset['tokens'].sum():,} tokens")

        # Verificar si hay documentos potencialmente duplicados
        unique_docs = documents_df['document'].nunique()
        total_docs = len(documents_df)
        if unique_docs < total_docs:
            print(f"\n📋 Análisis de duplicados:")
            print(f"   - Documentos únicos: {unique_docs}")
            print(f"   - Total de documentos: {total_docs}")
            print(f"   - Documentos duplicados: {total_docs - unique_docs}")

    return documents_df

def save_documents_dataframe(documents_df: pd.DataFrame, base_path: str):
    """
    Guarda el dataframe de documentos en diferentes formatos
    """
    if documents_df.empty:
        print("❌ No hay documentos para guardar")
        return

    try:
        # Guardar como pickle (con todas las columnas)
        pickle_path = f"{base_path}_documents_tokens.pkl"
        documents_df.to_pickle(pickle_path)
        print(f"💾 DataFrame completo guardado en: {pickle_path}")

        # Guardar como CSV (formato más legible)
        csv_path = f"{base_path}_documents_tokens.csv"
        documents_df.to_csv(csv_path, index=False)
        print(f"💾 DataFrame guardado como CSV en: {csv_path}")

        # Crear una versión resumida para Excel (sin el texto completo del documento)
        summary_df = documents_df[['document_id', 'supertopic', 'tokens', 'char_length']].copy()
        excel_path = f"{base_path}_documents_summary.xlsx"
        summary_df.to_excel(excel_path, index=False)
        print(f"💾 Resumen guardado como Excel en: {excel_path}")

        # Crear archivo con solo document y tokens como pediste
        simple_df = documents_df[['document', 'tokens']].copy()
        simple_csv_path = f"{base_path}_documents_tokens_simple.csv"
        simple_df.to_csv(simple_csv_path, index=False)
        print(f"💾 Archivo simple (document, tokens) guardado en: {simple_csv_path}")

    except Exception as e:
        print(f"❌ Error al guardar archivos: {str(e)}")

# Configuración con los datos proporcionados
questions_dict = {
    "Physical Layer Security in FSO Communication": [
        "Which modulation schemes are most secure against eavesdropping in atmospheric turbulence conditions?",
        "List comparative BER and secrecy capacity values for DPSK, PPM, and OAM modulation schemes under weak and strong turbulence.",
        "How do RIS-assisted FSO systems enhance secrecy capacity compared to conventional reflectors?",
        "What adaptive beamforming techniques are used to steer FSO beams securely toward legitimate users?",
        "How does adaptive beamforming reduce interception risk in hybrid RF-FSO systems?",
        "What is the role of channel state information (CSI) in implementing secure beamforming?",
        "Are there simulations showing secrecy rate improvement with beam steering under different turbulence conditions?",
        "What types of relays (amplify-and-forward, decode-and-forward) are used in secure FSO systems?",
        "How does cooperative relaying improve physical layer security under eavesdropping scenarios?",
        "What are the relay selection techniques that maximize secrecy capacity in mixed RF-FSO networks?",
        "What channel coding methods are suitable for secure FSO links?",
        "How do LDPC-based secrecy codes perform under strong turbulence fading?",
        "What are the design parameters for secrecy-capacity-achieving codes over fading optical channels?",
        "Can multilevel coding be used to separate data and confidential bits in a secure FSO transmitter?",
        "What are practical coding schemes that optimize secrecy throughput and BER simultaneously?",
        "How is spatial diversity implemented in FSO systems using MISO/MIMO architectures?",
        "What diversity combining techniques (e.g., selection, maximal ratio) are optimal for security?",
        "How does spatial diversity mitigate the effects of fading and improve secrecy rate in dual-hop RF-FSO networks?",
        "Can spatial diversity be combined with artificial noise to confuse eavesdroppers?"
    ],
    "Quantum Key Distribution (QKD) over FSO Systems": [
        "What quantum modulation techniques are most effective for key exchange over FSO links?",
        "Explain how CVQKD improves security in satellite-based FSO networks.",
        "How can quantum key distribution (QKD) be integrated with OAM modes in FSO links?",
        "What advantages does OAM offer for implementing high-dimensional quantum states in QKD?",
        "What are the experimental demonstrations of OAM-based quantum communication over satellite or long-range FSO links?",
        "How does mode fidelity impact the security of OAM-QKD systems?",
        "What are current limitations in combining CVQKD with OAM in practical FSO systems?"
    ],
    "Hybrid RF-FSO Security Strategies": [
        "Can you provide performance plots of RIS-assisted dual-hop RF-FSO networks under eavesdropping attacks?",
        "Can relays also act as friendly jammers to enhance secure communication in dual-hop systems?",
        "What is the impact of relay location and mobility (e.g., UAV relays) on secrecy performance?",
        "How does spatial diversity mitigate the effects of fading and improve secrecy rate in dual-hop RF-FSO networks?",
        "Are there comparisons of spatial diversity schemes in secure FSO vs. hybrid FSO-RF configurations?"
    ],
    "Advanced Modulation and Security Techniques": [
        "Explain how chaotic modulation improves physical layer security in FSO channels.",
        "What is orbital angular momentum and how is it generated in FSO systems?",
        "What are the major challenges of implementing OAM-based secure FSO links under atmospheric turbulence?",
        "Which optical devices are used for encoding and decoding OAM modes in secure communication systems?",
        "What is the principle of reconfigurable intelligent surfaces in optical and hybrid RF-FSO systems?",
        "How can RIS enhance secrecy capacity by manipulating signal reflection toward intended users?",
        "What are the optimal phase-shift design methods in RIS to nullify signal leakage to eavesdroppers?",
        "Can RIS be used to dynamically control the secure path in ground-to-satellite FSO communication?",
        "Are there practical RIS prototypes or testbeds for secure optical wireless systems?"
    ],
    "6G and Future Quantum Security Challenges": [
        "How can LSTM or AI models enhance adaptive beamforming in dynamic FSO environments?",
        "Integrating Quantum communication with OAM"
    ]
}

category_indices = {
    "Physical Layer Security in FSO Communication": [0, 2, 9, 13],
    "Quantum Key Distribution (QKD) over FSO Systems": [1, 7, 12],
    "Hybrid RF-FSO Security Strategies": [3, 4, 6, 8, 10, 11, 14],
    "Advanced Modulation and Security Techniques": [5, 10, 13],
    "6G and Future Quantum Security Challenges": [12]
}

#df_paths = {
#    'df_adv_modulation': '/content/drive/MyDrive/Phd(c)/REVIEW_SEGURIDAD/SUMMARIZATION BY BERTOPIC/df_adv_modulation.pkl',
#    'df_qkd': '/content/drive/MyDrive/Phd(c)/REVIEW_SEGURIDAD/SUMMARIZATION BY BERTOPIC/df_qkd.pkl',
#    'df_physical_layer': '/content/drive/MyDrive/Phd(c)/REVIEW_SEGURIDAD/SUMMARIZATION BY BERTOPIC/df_physical_layer.pkl',
#    'df_hybrid_rf_fso': '/content/drive/MyDrive/Phd(c)/REVIEW_SEGURIDAD/SUMMARIZATION BY BERTOPIC/df_hybrid_rf_fso.pkl',
#    'df_6g_quantum': '/content/drive/MyDrive/Phd(c)/REVIEW_SEGURIDAD/SUMMARIZATION BY BERTOPIC/df_6g_quantum.pkl'
#}

# Ejecutar la función principal
if __name__ == "__main__":
    # Crear el dataframe de supertópicos
    supertopic_df = create_supertopic_dataframe(questions_dict, category_indices, df_paths)

    # Mostrar información del resultado
    if not supertopic_df.empty:
        print("\n" + "="*50)
        print("RESUMEN DEL DATAFRAME DE SUPERTÓPICOS:")
        print("="*50)

        for idx, row in supertopic_df.iterrows():
            print(f"\n🔹 Supertópico: {row['supertopic']}")
            print(f"   📝 Preguntas: {row['num_questions']}")
            print(f"   📄 Documentos: {row['num_documents']}")

        # Crear DataFrame de documentos individuales con tokens
        print("\n" + "="*50)
        print("CREANDO DATAFRAME DE DOCUMENTOS INDIVIDUALES:")
        print("="*50)

        documents_df = create_documents_tokens_dataframe(supertopic_df)

        # Guardar ambos dataframes
        #base_path = '/content/drive/MyDrive/Phd(c)/REVIEW_SEGURIDAD/supertopic_dataframe'
        base_path = os.path.join(main_path, 'supertopic_dataframe')


        # Guardar dataframe de supertópicos
        try:
            supertopic_df.to_pickle(f"{base_path}.pkl")
            print(f"\n💾 DataFrame de supertópicos guardado en: {base_path}.pkl")
        except Exception as e:
            print(f"\n❌ Error al guardar supertópicos: {str(e)}")

        # Guardar CSV de supertópicos
        try:
            df_for_csv = supertopic_df.copy()
            df_for_csv['questions_str'] = df_for_csv['questions'].apply(lambda x: ' | '.join(x))
            df_for_csv['documents_str'] = df_for_csv['documents'].apply(lambda x: ' | '.join(x) if x else '')
            df_for_csv[['supertopic', 'questions_str', 'documents_str', 'num_questions', 'num_documents']].to_csv(f"{base_path}.csv", index=False)
            print(f"💾 DataFrame de supertópicos también guardado como CSV en: {base_path}.csv")
        except Exception as e:
            print(f"❌ Error al guardar CSV de supertópicos: {str(e)}")

        # Guardar dataframe de documentos
        if not documents_df.empty:
            save_documents_dataframe(documents_df, base_path)

    print("\nProceso completado!")
```
